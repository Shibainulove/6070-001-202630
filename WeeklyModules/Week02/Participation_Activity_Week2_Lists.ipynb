## Task 2: Creating a list and changing a value

```python
# 2a: create a variable called spam that contains [2, 4, 6, 8, 10]
spam = [2, 4, 6, 8, 10]
print('Before:', spam)

# 2b: make 'hello' the third value. List indexes start at 0,
# so the third value is at index 2.
spam[2] = 'hello'
print('After: ', spam)
```

## Task 3: Indexing and slicing

```python
# 3a: create a variable called letters
letters = ['a', 'b', 'c', 'd']

# 3b: letters[int(int('3'*2)//11)] evaluates to 'd'.
# Working from the inside out:
#   '3'*2       -> '33'  (multiplying a string by an integer repeats the string)
#   int('33')   -> 33    (converts the string to an integer)
#   33//11      -> 3     (// is floor division, which returns a whole number)
#   int(3)      -> 3
#   letters[3]  -> 'd'   (indexes start at 0, so index 3 is the fourth item)
print('3b:', letters[int(int('3'*2)//11)])

# 3c: letters[-1] evaluates to 'd'.
# Negative indexes count backward from the end of the list, so -1 is the last item.
print('3c:', letters[-1])

# 3d: letters[:2] evaluates to ['a', 'b'].
# A slice [start:end] returns a new list that goes up to, but does not include,
# the end index. Leaving out the start means "from the beginning", so this
# returns the items at indexes 0 and 1.
print('3d:', letters[:2])
```

## Task 4: Converting between lists and tuples

Passing a list to `tuple()` returns its tuple form, and passing a tuple to `list()` returns its list form.

```python
# tuple() gives the tuple form of a list
letters_tuple = tuple(letters)
print(letters_tuple, type(letters_tuple))

# list() gives the list form of a tuple
letters_list = list(letters_tuple)
print(letters_list, type(letters_list))

# Tuples are immutable (they can't be changed), so converting a tuple
# to a list is how you would add, remove or change its values
letters_list.append('e')
print(letters_list)
```

## Task 5: Comma Code practice project

A function that takes any list and returns its items as one string, separated by a comma and a space, with "and" before the last item.

```python
def comma_code(items):
    '''Return the items in a list as a string separated by commas, with 'and' before the last item.

    Works with any list: str() converts non-string values such as numbers,
    and an empty list returns an empty string.
    '''
    words = [str(item) for item in items]

    if len(words) == 0:
        return ''
    if len(words) == 1:
        return words[0]
    if len(words) == 2:
        return words[0] + ' and ' + words[1]  # two items don't need a comma

    # Join every item except the last with ', ', then add ', and ' and the last item
    return ', '.join(words[:-1]) + ', and ' + words[-1]


spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(spam))

# Test with other lists, including an empty list
print(repr(comma_code([])))
print(comma_code(['apples']))
print(comma_code(['apples', 'bananas']))
print(comma_code([1, 2, 3]))
```
