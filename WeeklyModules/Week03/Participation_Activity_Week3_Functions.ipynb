## Task 3: Questions about functions

### 3a. When does the code in a function execute: when the function is defined or called?

**Answer:** When the function is called. The `def` statement only creates the function and stores it under its name. The code inside the function does not run until a call such as `hello()` is made, and it runs again on every call.

```python
# The def statement only defines the function. Nothing inside it runs yet.
def hello():
    '''Print a line, to show when the code inside a function runs.'''
    print('   -> the code inside hello() is running')


print('1. hello() has been defined, and it has not printed anything')

# The code inside the function runs only when it is called, once for each call
print('2. Calling hello() the first time:')
hello()
print('3. Calling hello() the second time:')
hello()
```

### 3b. How many global scopes are there in a Python program? How many local scopes?

**Answer:** There is exactly one global scope. It is created when the program starts and destroyed when the program ends. There can be any number of local scopes, because a new one is created every time a function is called.

```python
eggs = 'global'  # assigned outside all functions, so it is in the one global scope


def bacon():
    '''Assign a local variable and print the names in this call's local scope.'''
    ham = 'local to bacon()'
    print('bacon() local scope:', locals())


def spam():
    '''Assign a local variable, print this call's local scope, then call bacon().'''
    eggs = 'local to spam()'  # a different variable from the global eggs
    print('spam() local scope: ', locals())
    bacon()  # this call gets its own local scope, separate from spam()'s


# Every call creates a new local scope: one for spam(), one for the bacon()
# call inside it, and one more for the bacon() call below
spam()
bacon()

# The local eggs inside spam() did not change the global eggs
print('The global eggs is still:', repr(eggs))
```

### 3c. What happens to variables in a local scope when the function call returns?

**Answer:** The local scope is destroyed and all the variables in it are forgotten. They can't be used from the global scope, and the next call to the function starts with a new local scope that does not remember the values from the earlier call. The way to keep a value is to return it.

```python
def count_call():
    '''Create a local counter, add 1 to it and return it.'''
    counter = 0  # created again on every call
    counter += 1
    return counter


# counter does not keep its value between calls, so both calls return 1
print('First call returns: ', count_call())
print('Second call returns:', count_call())

# After the call returns, the local variable counter no longer exists.
# Using it in the global scope raises a NameError, which is caught here
# so the rest of the notebook keeps running.
try:
    print(counter)
except NameError as error:
    print('NameError:', error)
```

### 3d. How can you prevent a program from crashing when it gets an error?

**Answer:** Put the line of code that might cause an error in a `try` clause, and add an `except` clause that says what to do if that error happens. When an error is raised inside the `try` clause, Python jumps to the `except` clause instead of stopping, and the program carries on from there.

```python
def divide_42_by(divide_by):
    '''Return 42 divided by divide_by, or None if divide_by is zero.'''
    try:
        return 42 / divide_by
    except ZeroDivisionError:
        # Dividing by zero raises a ZeroDivisionError. Without the try and
        # except clauses the program would stop here with an error message.
        print('Error: Invalid argument.')


print(divide_42_by(2))
print(divide_42_by(0))   # the error is handled, so the program keeps going
print(divide_42_by(12))
```

## Task 5: The Collatz Sequence practice program

### The `collatz()` function

`collatz()` takes one number and works out the next number in the Collatz sequence. If the number is even, it prints and returns `number // 2`. If the number is odd, it prints and returns `3 * number + 1`. The docstring at the top of the function documents what it does, its parameter and its return value.

```python
def collatz(number):
    '''Print and return the next number in the Collatz sequence.

    Parameters:
        number (int): the current number in the sequence.

    Returns:
        int: number // 2 if number is even, or 3 * number + 1 if number is odd.
    '''
    if number % 2 == 0:
        # Even: dividing by 2 leaves no remainder. The // operator is integer
        # division, so the result stays an integer instead of a float like 5.0.
        result = number // 2
    else:
        # Odd
        result = 3 * number + 1

    print(result)
    return result
```

```python
# Test the function on its own before using it in the program.
# 6 is even, so the next number is 6 // 2 = 3.
# 3 is odd, so the next number is 3 * 3 + 1 = 10.
next_number = collatz(6)
next_number = collatz(next_number)

# help() displays the docstring written inside the function
help(collatz)
```

### The program, with input validation

The program asks for a whole number and keeps calling `collatz()` on the result until the function returns 1.

`input()` always returns a string, so `int()` converts it to an integer. If the text is not a whole number (for example `puppy`), `int()` raises a `ValueError`. The `try` and `except` statements catch that error and ask again instead of letting the program crash. Zero and negative numbers are turned away too, because the sequence never reaches 1 from them and the loop would run forever.

Run the cell and type a number in the box that appears. The saved output below comes from typing `puppy`, then `0`, then `3`.

```python
# Keep asking until the user enters a positive integer
while True:
    try:
        # input() returns a string. int() raises a ValueError if the string
        # is not a whole number, such as 'puppy' or '2.5'.
        number = int(input('Enter number: '))
    except ValueError:
        print('You must enter an integer, such as 3.')
        continue  # go back to the top of the loop and ask again

    if number < 1:
        # From 0 or a negative number the sequence never reaches 1,
        # so the loop below would never end
        print('The number must be 1 or greater.')
        continue

    break  # the input is valid, so stop asking

# Keep calling collatz() on its own return value until it returns 1
while number != 1:
    number = collatz(number)
```
