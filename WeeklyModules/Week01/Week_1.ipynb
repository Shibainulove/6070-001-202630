**2. Are all data normally distributed?** No. The normal distribution is a useful model, but many real data sets don't follow it. Incomes, salaries and house prices are right-skewed, with a long tail of very high values. Counts, such as arrivals per hour, follow a Poisson distribution, and waiting times are often exponential. Some data have more than one peak (multimodal), and bounded measures such as percentages or test scores can't have the endless tails of a bell curve. Even the `salaries` generated in Exercise 4 aren't normal: `random.random()` draws from a *uniform* distribution, so values anywhere from 0 to 1,000,000 are equally likely. Normality has to be checked (with a histogram, a Q–Q plot or a test such as Shapiro–Wilk) rather than assumed. The Central Limit Theorem says that *sample means* become approximately normal as samples get larger, not that the data themselves are normal.

**3. When is the median a better measure of center than the mean?** When the data are skewed or contain outliers. The mean uses every value, so a few extreme values pull it toward the long tail, while the median depends only on the middle of the sorted data. Income is the classic example: a handful of very high earners push the mean well above what a typical person earns, which is why household income is usually reported as a median. The same goes for home prices, hospital stays or response times. The median also suits ordinal data, such as survey answers from "strongly disagree" to "strongly agree", where averaging the categories isn't really meaningful. For roughly symmetric data without outliers, the mean and median are close, and the mean is usually preferred because it uses all of the information.

## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
import math  # for math.isclose() and math.sqrt() -- not for any statistics

# By hand: add up all the salaries and divide by how many there are
n = len(salaries)
mean_salary = sum(salaries) / n

# Verify with the statistics module.
# math.isclose() is used instead of == because floating-point arithmetic
# can differ in the last decimal places depending on how a sum is computed.
import statistics
print(f'By hand:           {mean_salary:,.2f}')
print(f'statistics module: {statistics.mean(salaries):,.2f}')
print('Match:', math.isclose(mean_salary, statistics.mean(salaries)))
```

### median

```python
# By hand: sort the salaries and take the middle value. With an even number of
# values there are two middle values, so the median is their average.
sorted_salaries = sorted(salaries)  # sorted() makes a new list; `salaries` keeps its order
middle = n // 2
if n % 2 == 1:
    median_salary = sorted_salaries[middle]
else:
    median_salary = (sorted_salaries[middle - 1] + sorted_salaries[middle]) / 2

# Verify with the statistics module
print(f'By hand:           {median_salary:,.2f}')
print(f'statistics module: {statistics.median(salaries):,.2f}')
print('Match:', math.isclose(median_salary, statistics.median(salaries)))
```

### mode

```python
# By hand: count how many times each salary appears, then keep the salary
# (or salaries, if there's a tie) with the highest count
counts = {}
for salary in salaries:
    counts[salary] = counts.get(salary, 0) + 1

highest_count = max(counts.values())
modes = [salary for salary, count in counts.items() if count == highest_count]

# Verify with the statistics module. mode() returns the most common value;
# multimode() returns every value tied for most common.
print(f'By hand:           {modes} (appears {highest_count} times)')
print(f'statistics module: mode() = {statistics.mode(salaries)}, multimode() = {statistics.multimode(salaries)}')
print('Match:', modes == statistics.multimode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
# By hand: the average squared distance from the mean, dividing by n - 1 instead
# of n (Bessel's correction) because we're estimating from a sample.
# The result is in squared units (dollars squared).
sample_variance = sum((x - mean_salary) ** 2 for x in salaries) / (n - 1)

# Verify with the statistics module (variance() is the sample version; pvariance() divides by n)
print(f'By hand:           {sample_variance:,.2f}')
print(f'statistics module: {statistics.variance(salaries):,.2f}')
print('Match:', math.isclose(sample_variance, statistics.variance(salaries)))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
# By hand: the square root of the sample variance, which puts it back in dollars
sample_std = math.sqrt(sample_variance)

# Verify with the statistics module (stdev() is the sample version; pstdev() divides by n)
print(f'By hand:           {sample_std:,.2f}')
print(f'statistics module: {statistics.stdev(salaries):,.2f}')
print('Match:', math.isclose(sample_std, statistics.stdev(salaries)))
```

## Exercise 6: Calculating more statistics
### range

```python
# Range: the largest salary minus the smallest
salary_range = max(salaries) - min(salaries)
print(f'Range: {max(salaries):,.0f} - {min(salaries):,.0f} = {salary_range:,.0f}')
```

### coefficient of variation

```python
# Coefficient of variation: the standard deviation relative to the mean. It has no
# units, so it can compare the spread of data sets measured on different scales.
cv = sample_std / mean_salary

print(f'By hand:           {cv:.4f}')
print(f'statistics module: {statistics.stdev(salaries) / statistics.mean(salaries):.4f}')
print(f'The standard deviation is {cv:.1%} of the mean.')
```

### interquartile range

```python
def quantile(data, q):
    '''Return the q-th quantile of `data`, for q between 0 and 1.

    Finds the position of q in the sorted data and interpolates linearly between
    the two closest values. This is the default method in pandas and NumPy, and
    the same as statistics.quantiles(..., method='inclusive').
    '''
    ordered = sorted(data)
    position = (len(ordered) - 1) * q  # 0-based position, e.g. 24.75 for Q1 here
    below, above = math.floor(position), math.ceil(position)
    weight = position - below          # how far we are between the two neighbours
    return ordered[below] + (ordered[above] - ordered[below]) * weight

# Interquartile range: the spread of the middle 50% of the data
q1 = quantile(salaries, 0.25)
q3 = quantile(salaries, 0.75)
iqr = q3 - q1
print(f'Q1 = {q1:,.2f}, Q3 = {q3:,.2f}')

# Verify with the statistics module. method='inclusive' matches the method above;
# the default, method='exclusive', places the quartiles slightly differently.
check_q1, check_q2, check_q3 = statistics.quantiles(salaries, n=4, method='inclusive')
print(f'By hand:           {iqr:,.2f}')
print(f'statistics module: {check_q3 - check_q1:,.2f}')
print('Match:', math.isclose(iqr, check_q3 - check_q1))
```

### quartile coefficent of dispersion

```python
# Quartile coefficient of dispersion: half the IQR divided by the midpoint of Q1 and Q3,
# which simplifies to (Q3 - Q1) / (Q3 + Q1). Like the CV it has no units, but because
# it's built from quartiles it is much less sensitive to outliers.
qcd = (q3 - q1) / (q3 + q1)

print(f'By hand:           {qcd:.4f}')
print(f'statistics module: {(check_q3 - check_q1) / (check_q3 + check_q1):.4f}')
```

## Exercise 7: Scaling data
### min-max scaling

```python
# Min-max scaling: subtract the minimum and divide by the range,
# so the smallest salary becomes 0 and the largest becomes 1
min_salary = min(salaries)
min_max_scaled = [(x - min_salary) / salary_range for x in salaries]

print('First five:', [round(x, 4) for x in min_max_scaled[:5]])
print('Min:', min(min_max_scaled), '  Max:', max(min_max_scaled))
```

### standardizing

```python
# Standardizing (z-scores): subtract the mean and divide by the standard deviation,
# so each value says how many standard deviations it is from the mean
standardized = [(x - mean_salary) / sample_std for x in salaries]

print('First five:', [round(z, 4) for z in standardized[:5]])
print(f'Mean: {statistics.mean(standardized):.4f}   Standard deviation: {statistics.stdev(standardized):.4f}')
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
# Covariance between the min-max scaled and the standardized salaries: the average
# product of each variable's distance from its own mean. Dividing by n - 1 gives the
# sample covariance (the population version, E[(X - E[X])(Y - E[Y])], divides by n).
mean_mm = sum(min_max_scaled) / n
mean_z = sum(standardized) / n
covariance = sum(
    (x - mean_mm) * (z - mean_z) for x, z in zip(min_max_scaled, standardized)
) / (n - 1)

# Verify with the statistics module (covariance() needs Python 3.10+, which Colab has)
print(f'By hand:           {covariance:.6f}')
print(f'statistics module: {statistics.covariance(min_max_scaled, standardized):.6f}')
print('Match:', math.isclose(covariance, statistics.covariance(min_max_scaled, standardized)))
```

### Pearson correlation coefficient ($\rho$)

```python
# Pearson correlation coefficient: the covariance divided by the product of the two
# standard deviations, which rescales it to between -1 and 1
std_mm = math.sqrt(sum((x - mean_mm) ** 2 for x in min_max_scaled) / (n - 1))
std_z = math.sqrt(sum((z - mean_z) ** 2 for z in standardized) / (n - 1))
pearson = covariance / (std_mm * std_z)

# Verify with the statistics module (correlation() needs Python 3.10+)
print(f'By hand:           {pearson:.6f}')
print(f'statistics module: {statistics.correlation(min_max_scaled, standardized):.6f}')
print('Match:', math.isclose(pearson, statistics.correlation(min_max_scaled, standardized)))
```

**Why is ρ exactly 1?** Min-max scaling and standardizing are both *linear* transformations of the same salaries: each one subtracts a constant and divides by a positive constant. Plotting one against the other therefore gives a perfectly straight, upward-sloping line, which is a perfect positive correlation. Scaling changes the units, not the shape or the order of the data. The covariance on its own (about 0.27) is harder to interpret because it depends on each variable's scale; dividing by the two standard deviations is what makes ρ comparable across data sets. Because z-scores have a standard deviation of 1 and ρ = 1, the covariance here equals the standard deviation of the min-max scaled data.
